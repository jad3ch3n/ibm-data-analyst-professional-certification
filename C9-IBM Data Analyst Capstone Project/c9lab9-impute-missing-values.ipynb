### Step 1: Import Required Libraries

### Step 2: Load the Dataset Into a Dataframe


#### **Read Data**

```python
file_path ="https://cf-courses-data.s3.us.cloud-object-storage.appdomain.cloud/UDKAZw-kz18Yj8P6icf_qw/survey-data-duplicates.csv"
df = pd.read_csv(file_path)

# Display the first few rows to ensure it loaded correctly
print(df.head())
```

### Step 3. Finding and Removing Duplicates
##### Task 1: Identify duplicate rows in the dataset.


```python
df.shape
```

```python
# Identify duplicate rows in the dataset
duplicate_rows = df[df.duplicated()]
print(f"Number of duplicate rows: {duplicate_rows.shape[0]}")
```

```python
duplicate_rows
```

##### Task 2: Remove the duplicate rows from the dataframe.



```python
# Remove the duplicate rows from the dataset
df = df.drop_duplicates()
print(f"Number of rows after removing duplicates: {df.shape[0]}")
```

### Step 4: Finding Missing Values
##### Task 3: Find the missing values for all columns.


```python
# Find the missing values for all columns
missing_values = df.isnull().sum().sort_values(ascending=False)
print("Missing values in each column:")
print(missing_values)
```

##### Task 4: Find out how many rows are missing in the column RemoteWork.


```python
# Count the number of missing rows for RemoteWork column
missing_remote_work = df['RemoteWork'].isnull().sum()
print(f"Number of missing values in RemoteWork column: {missing_remote_work}")
```

### Step 5. Imputing Missing Values
##### Task 5: Find the value counts for the column RemoteWork.


```python
# Find the value counts for the column RemoteWork
remote_work_counts = df['RemoteWork'].value_counts()
print("Value counts for RemoteWork column:")
print(remote_work_counts)
```

##### Task 6: Identify the most frequent (majority) value in the RemoteWork column.



```python
# Identify the most frequent value in the RemoteWork column
most_frequent_remote_work = df['RemoteWork'].mode()[0]
print(f"The most frequent value in RemoteWork column: {most_frequent_remote_work}")
```

##### Task 7: Impute (replace) all the empty rows in the column RemoteWork with the majority value.



```python
# Inpute missing values in Employment column with the most frequent value
df['RemoteWork'] = df['RemoteWork'].fillna(most_frequent_remote_work)
print(f"Number of missing values in RemoteWork column after imputation: {df['RemoteWork'].isnull().sum()}")
```
