### 1. Load the Dataset


<h5>1.1 Import necessary libraries and load the dataset.</h5>


Ensure the dataset is loaded correctly by displaying the first few rows.


```python
# Import necessary libraries
import pandas as pd

# Load the Stack Overflow survey data
dataset_url = "https://cf-courses-data.s3.us.cloud-object-storage.appdomain.cloud/n01PQ9pSmiRX6520flujwQ/survey-data.csv"
df = pd.read_csv(dataset_url)

# Display the first few rows
print(df.head())
```

#### 2. Explore the Dataset


<h5>2.1 Summarize the dataset by displaying the column data types, counts, and missing values.</h5>


```python
summary = pd.DataFrame({
    "Data Type": df.dtypes,
    "Non-Null Count": df.count(),
    "Missing Values": df.isnull().sum()
})

print(summary)
```

<h5>2.2 Generate basic statistics for numerical columns.</h5>


```python
df.describe()
```

### 3. Identifying and Removing Inconsistencies


<h5>3.1 Identify inconsistent or irrelevant entries in specific columns (e.g., Country).</h5>


```python
invalid_entries = df[
    df["Country"].isnull()]
invalid_entries["Country"]
```

```python
len(invalid_entries[["Country"]])
```

```python
len(df)
```

```python
df[['Country']].value_counts().head(20)
```

<h5>3.2 Standardize entries in columns like Country or EdLevel by mapping inconsistent values to a consistent format.</h5>


```python
df = df.dropna(subset=["Country", "EdLevel"])
df["Country"].value_counts(dropna=False)
df["EdLevel"].value_counts(dropna=False)
```

### 4. Encoding Categorical Variables


<h5>4.1 Encode the Employment column using one-hot encoding.</h5>


```python
df = pd.get_dummies(df, columns=["Employment"])
df.head()
```

### 5. Handling Missing Values


<h5>5.1 Identify columns with the highest number of missing values.</h5>


```python
missing_counts = df.isnull().sum().sort_values(ascending=False)
print(missing_counts)
```

<h5>5.2 Impute missing values in numerical columns (e.g., `ConvertedCompYearly`) with the mean or median.</h5>


```python
num_cols = df.select_dtypes(include="number").columns
df[num_cols] = df[num_cols].fillna(df[num_cols].median())
if df[num_cols].isnull().sum().all() == 0:
    print("All missing values in numerical columns have been inputed with the median.")
```

```python
df["ConvertedCompYearly"] = df["ConvertedCompYearly"].fillna(
    df["ConvertedCompYearly"].median()
)
df['ConvertedCompYearly'].head(100)
```

<h5>5.3 Impute missing values in categorical columns (e.g., `RemoteWork`) with the most frequent value.</h5>


```python
df[['RemoteWork']].dtypes
```

```python
cat_cols = df.select_dtypes(include=["str", "object", "category"]).columns
for col in cat_cols:
    df[col] = df[col].fillna(df[col].mode()[0])
if df[cat_cols].isnull().sum().all() == 0:
    print("All missing values in categorical columns have been inputed with the mode.")
```

### 6. Feature Scaling and Transformation


<h5>6.1 Apply Min-Max Scaling to normalize the `ConvertedCompYearly` column.</h5>


```python
# Check that ConvertedCompYearly does not have missing values
print(df['ConvertedCompYearly'].isnull().sum())
```

```python
# compute min and max of the column
min_val = df['ConvertedCompYearly'].min()
max_val = df['ConvertedCompYearly'].max()

# apply Min-Max scaling
df['ConvertedCompYearly'] = (
    df['ConvertedCompYearly'] - min_val
) / (max_val - min_val)
```

```python
df.head(2)
```

<h5>6.2 Log-transform the ConvertedCompYearly column to reduce skewness.</h5>


```python
print(df["ConvertedCompYearly"].skew())
```

```python
import numpy as np
df["ConvertedCompYearly"] = np.log1p(df["ConvertedCompYearly"])
```

```python
print(df["ConvertedCompYearly"].skew())
```

### 7. Feature Engineering


<h5>7.1 Create a new column `ExperienceLevel` based on the `YearsCodePro` column:</h5>


```python
df['YearsCodePro'].value_counts(dropna=False).index
```

```python
df["YearsCodePro"] = df["YearsCodePro"].replace({
    "Less than 1 year": 0,
    "More than 50 years": 51
})

df["YearsCodePro"] = pd.to_numeric(df["YearsCodePro"], errors="coerce")
df["YearsCodePro"].dtype
```

```python
def get_experience_level(years):
    if pd.isna(years):
        return "Unknown"
    elif years < 2:
        return "Junior"
    elif years < 5:
        return "Mid"
    elif years < 10:
        return "Senior"
    else:
        return "Expert"

df["ExperienceLevel"] = df["YearsCodePro"].apply(get_experience_level)
```
